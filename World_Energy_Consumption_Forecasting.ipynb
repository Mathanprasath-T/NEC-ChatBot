```python

df.columns = df.columns.str.strip()

print("Available columns:")
for col in df.columns:
    print(repr(col))
```

```python

# Select India
india = df[df['Country'] == 'India'].copy()

print(india.head())
print("Number of rows:", len(india))
```

```python

# Convert year columns into rows
india_long = india.melt(
    id_vars=['Country'],
    var_name='Year',
    value_name='Energy_Consumption'
)

# Convert data types
india_long['Year'] = pd.to_numeric(
    india_long['Year'], errors='coerce'
)

india_long['Energy_Consumption'] = pd.to_numeric(
    india_long['Energy_Consumption'], errors='coerce'
)

# Handle missing values and sort by year
india_long = india_long.dropna()
india_long = india_long.sort_values('Year')

print(india_long.head())
print(india_long.tail())
print("Total records:", len(india_long))
```

```python

import numpy as np
from sklearn.preprocessing import MinMaxScaler

values = india_long[
    ['Energy_Consumption']
].values

split = int(len(values) * 0.8)

scaler = MinMaxScaler()
scaler.fit(values[:split])

scaled_values = scaler.transform(values)

print("Normalization completed!")
print("Total years:", len(scaled_values))
```

```python

X = []
y = []

for i in range(5, len(scaled_values)):
    X.append(scaled_values[i-5:i])
    y.append(scaled_values[i])

X = np.array(X)
y = np.array(y)

split_seq = split - 5

X_train = X[:split_seq]
y_train = y[:split_seq]

X_test = X[split_seq:]
y_test = y[split_seq:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

model = Sequential([
    LSTM(32, input_shape=(5, 1)),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.summary()
```

```python

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=8,
    validation_split=0.2,
    shuffle=False,
    verbose=1
)
```

```python

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error

predictions = model.predict(X_test)

actual = scaler.inverse_transform(y_test)
predicted = scaler.inverse_transform(predictions)

mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))

print("MAE:", mae)
print("RMSE:", rmse)
```

```python

import matplotlib.pyplot as plt

test_years = india_long['Year'].iloc[split:]

plt.figure(figsize=(10, 5))
plt.plot(
    test_years, actual.flatten(),
    marker='o', label='Actual'
)
plt.plot(
    test_years, predicted.flatten(),
    marker='x', label='Predicted'
)

plt.xlabel('Year')
plt.ylabel('Energy Consumption')
plt.title('India Energy Consumption Forecasting')
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Get the latest 5 years of data
last_five_years = scaled_values[-5:]

# Reshape data for LSTM
last_five_years = last_five_years.reshape(1, 5, 1)

# Predict next year
next_year_scaled = model.predict(last_five_years)

# Convert prediction to original scale
next_year_prediction = scaler.inverse_transform(
    next_year_scaled
)

last_year = int(india_long['Year'].iloc[-1])

print("Last recorded year:", last_year)
print("Predicted year:", last_year + 1)
print(
    "Predicted energy consumption:",
    round(float(next_year_prediction[0][0]), 3)
)
```
